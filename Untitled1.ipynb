```python

import pandas as pd

# Load original dataset
df = pd.read_csv("Exam_Score_Prediction.csv")

print("Original Dataset Shape:", df.shape)

# Remove duplicate records
df = df.drop_duplicates()

# Remove rows where target score is missing
df = df.dropna(subset=["exam_score"])

# Handle missing numerical values
numerical_columns = df.select_dtypes(
    include=["int64", "float64"]
).columns

for col in numerical_columns:
    df[col] = df[col].fillna(df[col].median())

# Handle missing categorical values
categorical_columns = df.select_dtypes(
    include=["object"]
).columns

for col in categorical_columns:
    if df[col].isnull().any():
        df[col] = df[col].fillna(df[col].mode()[0])

# Save cleaned dataset
df.to_csv("Exam_Score_Prediction_Cleaned.csv", index=False)

print("\nData Cleaning Completed!")
print("Cleaned Dataset Shape:", df.shape)
print("\nMissing Values:")
print(df.isnull().sum())

print("\nCleaned CSV file saved successfully!")
```

```python

import pandas as pd
import numpy as np

# ====================================
# 1. LOAD DATASET
# ====================================

df = pd.read_csv("Exam_Score_Prediction_Cleaned.csv")

print("Dataset Loaded Successfully!")
print("Total Students:", len(df))


# ====================================
# 2. PREPARE DATA
# ====================================

# Remove unnecessary student ID
df = df.drop(columns=["student_id"])

# Separate input and output
X = df.drop(columns=["exam_score"])
y = df["exam_score"].to_numpy()

# Convert categorical data into numerical data
X = pd.get_dummies(X, dtype=float)

feature_columns = X.columns.tolist()

X = X.to_numpy(dtype=float)


# ====================================
# 3. TRAIN AND TEST SPLIT
# ====================================

np.random.seed(42)

indices = np.random.permutation(len(X))

split = int(0.8 * len(X))

train_indices = indices[:split]
test_indices = indices[split:]

X_train = X[train_indices]
X_test = X[test_indices]

y_train = y[train_indices]
y_test = y[test_indices]


# ====================================
# 4. NORMALIZE DATA
# ====================================

mean = X_train.mean(axis=0)
std = X_train.std(axis=0)

# Prevent division by zero
std[std == 0] = 1

X_train = (X_train - mean) / std
X_test = (X_test - mean) / std


# ====================================
# 5. TRAIN LINEAR REGRESSION
# ====================================

# Add intercept column
X_train_b = np.c_[
    np.ones(len(X_train)), X_train
]

X_test_b = np.c_[
    np.ones(len(X_test)), X_test
]

# Calculate model coefficients
weights = np.linalg.lstsq(
    X_train_b,
    y_train,
    rcond=None
)[0]

print("\nModel Training Completed!")


# ====================================
# 6. TEST MODEL PERFORMANCE
# ====================================

predictions = X_test_b @ weights

# Mean Absolute Error
mae = np.mean(np.abs(y_test - predictions))

# Root Mean Squared Error
rmse = np.sqrt(np.mean((y_test - predictions) ** 2))

# R2 Score
ss_res = np.sum((y_test - predictions) ** 2)
ss_total = np.sum((y_test - np.mean(y_test)) ** 2)

r2 = 1 - (ss_res / ss_total)

print("\n========== MODEL PERFORMANCE ==========")

print("Mean Absolute Error:", round(mae, 2))
print("Root Mean Squared Error:", round(rmse, 2))
print("R2 Score:", round(r2, 4))
print("R2 Percentage:", round(r2 * 100, 2), "%")


# ====================================
# 7. TAKE STUDENT INPUT
# ====================================

print("\n================================")
print("       EXAM SCORE PREDICTOR")
print("================================")

print("\nEnter Student Details:\n")


def get_number(message, minimum, maximum):

    while True:
        try:
            value = float(input(message))

            if minimum <= value <= maximum:
                return value

            print("Enter value between", minimum, "and", maximum)

        except ValueError:
            print("Please enter a valid number.")


def get_choice(message, choices):

    print("\n" + message)

    for choice in choices:
        print("-", choice)

    while True:
        value = input("Enter choice: ").strip().lower()

        if value in choices:
            return value

        print("Invalid choice. Try again.")


# Numerical inputs

age = get_number("Enter Age (17-24): ", 17, 24)

study_hours = get_number(
    "Enter Daily Study Hours (0.08-7.91): ", 0.08, 7.91
)

attendance = get_number(
    "Enter Attendance Percentage (40.6-99.4): ", 40.6, 99.4
)

sleep_hours = get_number(
    "Enter Sleep Hours (4.1-9.9): ", 4.1, 9.9
)


# Categorical inputs

gender = get_choice(
    "Select Gender:",
    ["male", "female", "other"]
)

course = get_choice(
    "Select Course:",
    ["b.com", "b.sc", "b.tech", "ba",
     "bba", "bca", "diploma"]
)

internet = get_choice(
    "Internet Access:",
    ["yes", "no"]
)

sleep_quality = get_choice(
    "Sleep Quality:",
    ["good", "average", "poor"]
)

study_method = get_choice(
    "Study Method:",
    ["coaching", "group study", "mixed",
     "online videos", "self-study"]
)

facility = get_choice(
    "Facility Rating:",
    ["high", "medium", "low"]
)

difficulty = get_choice(
    "Exam Difficulty:",
    ["easy", "moderate", "hard"]
)


# ====================================
# 8. CREATE STUDENT DATA
# ====================================

student = pd.DataFrame([{
    "age": age,
    "gender": gender,
    "course": course,
    "study_hours": study_hours,
    "class_attendance": attendance,
    "internet_access": internet,
    "sleep_hours": sleep_hours,
    "sleep_quality": sleep_quality,
    "study_method": study_method,
    "facility_rating": facility,
    "exam_difficulty": difficulty
}])


# Convert categorical values
student = pd.get_dummies(student, dtype=float)

# Match training columns
student = student.reindex(
    columns=feature_columns,
    fill_value=0
)

# Normalize student data
student = (student.to_numpy() - mean) / std


# ====================================
# 9. PREDICT EXAM SCORE
# ====================================

student_b = np.c_[np.ones(1), student]

predicted_score = student_b @ weights

predicted_score = np.clip(
    predicted_score[0], 0, 100
)

# ====================================
# 10. DISPLAY RESULT
# ====================================

print("\n================================")
print("         PREDICTION RESULT")
print("================================")

print("Predicted Exam Score:",
      round(predicted_score, 2), "/ 100")

if predicted_score >= 80:
    print("Performance: Excellent")

elif predicted_score >= 65:
    print("Performance: Good")

elif predicted_score >= 50:
    print("Performance: Average")

elif predicted_score >= 35:
    print("Performance: Below Average")

else:
    print("Performance: Needs Improvement")

print("\nThank You for Using Exam Score Predictor!")
print("================================")
```
